```python
skills = ["python", "sql", "machine learning", "excel", "communication", "data analysis"]

resume = "python excel communication".lower()

found_skills = list(filter(lambda s: s in resume, skills))
missing_skills = list(filter(lambda s: s not in resume, skills))

print("Detected Skills:", found_skills)
print("Missing Skills:", missing_skills)

score = int((len(found_skills) / len(skills)) * 100)
print("Resume Score:", score, "%")

print("Suggestion: Try learning " + ", ".join(missing_skills)) if missing_skills else print("Great! Your resume is strong.")
```
